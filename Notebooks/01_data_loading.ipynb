Duplicate rows(training examples) will increase the weight for those particular examples during training which is not bad for the model , hence we will drop all the duplicates

```python
print(df.shape[0])
df.drop_duplicates(inplace=True) # this will delete all the duplicate rows while only keeping their first occurance
print(df.duplicated().sum()) # to confirm that u have successfully removed those rows
print(df.shape[0]) # should be 167 less than before bcuz previously we had 167 duplicate rows
```
